```python
#Viết hàm tìm kiếm nhị phântrên một danh sách đãđược sắp xếp tăng dần.
#Trả về chỉ số của phần tử hoặc -1 nếu không tìmthấy. mẫu hàm: def
#binary_search(arr, left,right, target):
def binary_search(arr,left,right,target):
    left=0
    right=len(arr)-1
    while(right>=left):
        mid=((left+right)//2)
        if int(arr[mid]) ==target :
            return mid
        elif int (arr[mid]) > target:
            right=mid-1
        else:
            left=mid+1
    return "-1"
n=list(map(int,input().split()))
t=int(input())
l=0
r=len(n)-1
print(binary_search(n,l,r,t))

            
```

```python
#Cho danh sách arr và mộtsố nguyên x, đếm xem xuất hiện bao nhiêu lần
#trong danh sách. def count_occurrences(arr, x) -> int:
def tim1(arr, x):
    arr.sort()
    l=0
    r=len(arr)-1
    res=-1
    while r>=l:
        mid=(l+r)//2
        if arr[mid] >= x :
            res=mid
            r=mid-1
        else :
            l=mid+1
    return res
def tim2(arr, x):
    arr.sort()
    l=0
    r=len(arr)-1
    res=-1
    while r>=l:
        mid=(l+r)//2
        if arr[mid] <= x :
            res=mid
            l=mid+1
        else :
            r=mid-1
    return res
arr=list(map(int,input().split()))
x=int(input())
s1=tim1(arr,x)
s2=tim2(arr,x)    
if s1==-1 or s2==-1:
    print(0)
else:
    print(s2-s1+1)
```
